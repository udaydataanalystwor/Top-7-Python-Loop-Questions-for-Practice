```python
def myname(dict1, dict2):
  total_products = {}

  for product, quantity in dict1.items():
    total_products[product] = quantity

  for product, quantity in dict2.items():
    if product in total_products:
      total_products[product] += quantity
    else:
      total_products[product] = quantity

  print("Total quantity of products:", total_products)


tech_products1 = {"laptop": 10, "charger": 15, "mobile": 5}
tech_products2 = {"laptop": 5, "mobile": 20, "charger": 12}

myname(tech_products1, tech_products2)
```
